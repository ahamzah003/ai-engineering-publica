## Task 1: Delivery Fee Calculator (Easy)

Ask for the order total. Delivery is **free** at ₦10000 or more, costs **₦500** at ₦5000 or more, and **₦1000** otherwise. Print the delivery fee and the total to pay.

**Expected Output** (example: order total `7500`):
```
Delivery fee : 500
Total to pay : 8000.0
```

````python
# ask for the order total because the delivery fee depends on it.
order_total = float(input("Enter your order total: "))

# decide the delivery fee based on thresholds.
# using if/elif/else ensures only one condition applies.
if order_total >= 10000:
    delivery_fee = 0
elif order_total >= 5000:
    delivery_fee = 500
else:
    delivery_fee = 1000

# calculate the final amount to pay by adding the fee.
total_to_pay = order_total + delivery_fee

# show the results clearly so the customer knows both parts.
print("\nDelivery fee :", delivery_fee)
print("Total to pay :", total_to_pay)

````

## Task 2: Till Total (Easy)

Use a `while` loop to keep asking for item prices until the cashier enters `0`. Skip negative prices (`continue`). Then print how many items there were and the total, and use a `for` loop with `range()` to print one `*` per item.

**Expected Output** (example: prices `1500`, `800`, `-20`, `2200`, then `0`):
```
Price cannot be negative.
Items: 3
Total: 4500.0
Basket: ***
```

````python
# keep track of both the number of items and the total cost
items = 0
total = 0.0

# loop keeps running until the cashier signals the end by typing 0.
while True:
    price = float(input("Enter item price (0 to finish): "))

    # 0 means "stop"
    if price == 0:
        break

    # skip negative prices
    if price < 0:
        print("Price cannot be negative.")
        continue

    # valid prices only should count towards the basket.
    items += 1
    total += price

# show the summary.
print("\nItems:", items)
print("Total:", total)

# prints one '*' per item to visualize the basket size.
print("Basket: ", end="")
for i in range(items):
    print("*", end="")
````

## Task 3: Name Tag (Medium)

Ask for a name that may have extra spaces and wrong capitals. Clean it with `.strip()` and `.title()`, then print a greeting using an **f-string** (you can use f-strings from this task on), the name in uppercase, its length, its first letter, and the name reversed.

**Expected Output** (example: name `"  ngozi okonkwo "`):
```
Welcome, Ngozi Okonkwo!
Uppercase   : NGOZI OKONKWO
Length      : 13
First letter: N
Reversed    : owknokO izogN
```

````python
# allows the user to provide any raw text.
raw_name = input("Enter your name: ")

# remove unwanted spaces at the start and end with .strip() 
# fix capitalization to proper case with .title() 
name = raw_name.strip().title()

# send greetings
print(f"Welcome, {name}!")

# show name as uppercase
print("Uppercase   :", name.upper())

# show number of characters in the name,
print("Length      :", len(name))

# show the first character of the name.
print("First letter:", name[0])

# show the name backwards.
print("Reversed    :", name[::-1])
````

## Task 4: Shopping List (Medium)

Start with the list `["Rice", "Beans", "Garri"]` and `.append()` `"Sugar"`. Ask which item to remove (clean it with `.strip().title()`); remove it if it is on the list, otherwise print a message. Use `enumerate()` to print the numbered list, then print how many items are left. Finally, unpack the tuple `("Mama Put Store", "Lagos")` into two variables and print them.

**Expected Output** (example: the user removes `"  beans "`):
```
1. Rice
2. Garri
3. Sugar
Total items: 3
Shop: Mama Put Store, Lagos
```

````python
# start with a predefined list of items
shopping_list = ["Rice", "Beans", "Garri"]

# add new item
shopping_list.append("Sugar")

# remove unwanted spaces at the start and end with .strip() 
# fix capitalization to proper case with .title() 
remove_item = input("Which item do you want to remove? ").strip().title()

# check membership before removing to avoid errors.
if remove_item in shopping_list:
    shopping_list.remove(remove_item)
else:
    print(f"{remove_item} is not in the list.\n")

# give both value and index starting from 1.
for index, item in enumerate(shopping_list, start=1):
    print(f"{index}. {item}")

# show how many items remain
print("Total items:", len(shopping_list))

# aasign values separately
shop_name, shop_location = ("Mama Put Store", "Lagos")
# show output
print(f"Shop: {shop_name}, {shop_location}")
````

## Task 5: Shopping Cart (Challenge)

Keep asking for an item name until the user types `done`. For each item, ask for its price and `.append()` a `(name, price)` tuple to a `cart` list. Then print a numbered list of the items with their prices, the total, and whether delivery is free (total of ₦10000 or more) or costs ₦1000.

**Expected Output** (example: `rice` at `6500`, `beans` at `3000`, `palm oil` at `2500`, then `done`):
```
1. Rice       ₦6500.00
2. Beans      ₦3000.00
3. Palm Oil   ₦2500.00
Total: ₦12000.00
Free delivery!
```

````python
## a list to keep both pieces of info together as tuple i.e., (name, price).
cart = []

# keeps running until the user types "done".
while True:
    # get name of item
    # remove unwanted spaces at the start and end with .strip() 
    # fix capitalization to proper case with .title()
    item_name = input("Enter item name (or 'done' to finish): ").strip().title()

    # "done" means stop signal, heance break out of the loop.
    if item_name.lower() == "done":
        break

    # get the price separately to pair it with the item.
    price = float(input(f"Enter price for {item_name}: "))

    # skip negative prices
    if price < 0:
        print("Price cannot be negative.")
        continue

    # keep each item's details together.
    cart.append((item_name, price))

# total is initially zero
total = 0.0


for index, (name, price) in enumerate(cart, start=1):
    print(f"{index}. {name}\t₦{round(price, 2)}")
    total += price

# show the total cost of the cart.
print(f"Total: ₦{round(total, 2)}")

# decide the delivery fee based on thresholds.
# using if/elif/else ensures only one condition applies.
if total >= 10000:
    delivery_fee = 0
else:
    delivery_fee = 1000

# show whether the delivery is free or has a cost
print("Free delivery!" if delivery_fee == 0 else f"Delivery fee is ₦{delivery_fee}.")
````
