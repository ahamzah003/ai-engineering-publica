## Task 3: Grade & Eligibility Calculator (Medium)

*Unlocks: + Topic 4 -- Operators & Expressions*

Ask for the student's **exam score**, **coursework score**, **attendance percentage** (all as floats), and **total study minutes** (as an int). Then:
1. Calculate `final_score` = exam score × 60% + coursework score × 40%
2. Give a flat 2-point participation bonus by adding it to `final_score` with `+=`
3. Compute `is_pass` (`final_score >= 50`) and `is_distinction` (`final_score >= 90` **and** attendance `>= 95`) and print them directly -- **do not use `if` yet**, that's next task
4. Using `//` and `%`, calculate how many complete 45-minute study blocks fit into the total study minutes, and how many minutes are left over

**Expected Output** (example: exam `78`, coursework `85`, attendance `92`, study minutes `300`):
```
Exam score           : 78.0
Coursework score     : 85.0
Attendance           : 92.0 %
Final score (+bonus) : 82.8

Is pass (>=50)       : True
Is distinction       : False

Total study minutes  : 300
Study blocks (45 min): 6
Minutes left over    : 30
```

````python
# collect student exam score
exam_score = float(input("Enter your exam score: "))
# collect student coursework score
cw_score = float(input("Enter your coursework score: "))
# collect student attendance score
attendance_pct = float(input("Enter your attendance percentage score: "))
# collect student total study minutes
total_study_minutes = int(input("Enter your total study duration(in minutes): "))

# compute the final score using both the exam score and coursework score
final_score = exam_score * 0.6 + cw_score * 0.4
# flat 2-point participation bonus
final_score += 2

# output
print("Exam score :", exam_score)
print("Coursework score :", cw_score)
print("Attendance : ", attendance_pct, "%", sep="")
print("Final score (+bonus) :", final_score, end="\n\n")

print("Is pass (>=50) :", final_score >= 50)
print("Is distinction :", final_score >= 90 and attendance_pct >= 95, end="\n\n")

print("Total study minutes :", total_study_minutes)
print("Study blocks (45 min) :", total_study_minutes // 45)
print("Minutes left over :", total_study_minutes % 45)
````

## Task 4: Grade Classifier (Medium)

*Unlocks: + Topic 5 -- Control Flow: Conditionals*

Reuse the score collection and `final_score` calculation from Task 3 (without the bonus). Then:
1. Ask for the student's name, but allow the user to just press Enter to skip it. Use a truthy/falsy check (`if not name:`) -- if no name was entered, use `"Anonymous Student"` instead
2. Using an `if` / `elif` / `else` chain (checked top to bottom): `final_score >= 90` → Grade `A`; `>= 80` → `B`; `>= 70` → `C`; `>= 60` → `D`; otherwise → `F`
3. Decide PASS/FAIL: PASS if `final_score >= 50` **and** attendance `>= 75`, otherwise FAIL
4. Print a formatted grade report with the name, final score, grade, and result

**Expected Output** (example: name skipped, exam `78`, coursework `85`, attendance `92`):
```
=== Grade Report ===
Student      : Anonymous Student
Final Score  : 80.8
Grade        : B
Result       : PASS
```

````python
# display username on the introduction card
user_name = input("Enter your fullname: ")
# collect student exam score
exam_score = float(input("Enter your exam score: "))
# collect student coursework score
cw_score = float(input("Enter your coursework score: "))
# collect student attendance score
attendance_pct = float(input("Enter your attendance percentage score: "))
# collect student total study minutes
total_study_minutes = int(input("Enter your total study duration(in minutes): "))

# compute the final score using both the exam score and coursework score
final_score = exam_score * 0.6 + cw_score * 0.4

if not user_name:
    user_name = "Anonymous Student"

# Determine grade using if/elif/else chain
if final_score >= 90:
    grade = "A"
elif final_score >= 80:
    grade = "B"
elif final_score >= 70:
    grade = "C"
elif final_score >= 60:
    grade = "D"
else:
    grade = "F"

# Determine PASS/FAIL
if final_score >= 50 and attendance_pct >= 75:
    result = "PASS"
else:
    result = "FAIL"

# output
# Print formatted grade report
print("="*3, "Grade Report", "="*3)
print("Student      :", user_name)
print("Final Score  :", final_score)
print("Grade        :", grade)
print("Result       :", result)
````

## Task 5: Class Grade Report Generator (Challenge -- Capstone)

*Unlocks: + Topic 6 -- Control Flow: Loops*

This is the Week 1 capstone -- turn your single-student grade classifier into a multi-student class report generator, using everything from Topics 1-6. Because lists aren't allowed until Week 2, report each student immediately as you go, and keep running totals instead of storing records.

1. Use a `while` loop to repeatedly collect a student's name. If the name is `"done"` (case-insensitive), stop collecting (`break`)
2. Otherwise, ask for exam score and coursework score. For attendance, use a `for attempt in range(3):` loop that re-asks up to 3 times if a negative number is entered, `break`-ing out as soon as a valid (non-negative) value is given
3. If all 3 attendance attempts were invalid, print a message and `continue` to skip this student entirely
4. Otherwise, calculate `final_score` and classify grade + PASS/FAIL exactly as in Task 4, then immediately print that student's numbered report line
5. Keep running totals with `+=`: a student counter, the sum of all final scores, and a count of students who passed
6. After the `while` loop ends, print the class average (total ÷ count) and the pass count (e.g., `"1 / 2"`)

**Expected Output** (example: Amara -- exam 95, coursework 95, attendance entered as `-5` then corrected to `95`; Kofi -- exam 40, coursework 55, attendance 60; then `"done"`):
```
--- Add Student (type 'done' as the name to finish) ---
  Invalid attendance -- try again.
1. Amara | Final: 95.0 | Grade: A | PASS
2. Kofi | Final: 46.0 | Grade: F | FAIL
========================================
Class average  : 70.5
Students passed: 1 / 2
========================================
```

````python
# Week 1 Capstone: Multi-student class report generator

student_counter = 0
sum_final_score = 0.0
pass_counter = 0

while True:
    print("\n--- Add Student (type 'done' as the name to finish) ---")
    user_name = input("Enter your fullname: ").strip()
    if user_name.lower() == "done":
        break

    exam_score = float(input("Enter your exam score: "))
    cw_score = float(input("Enter your coursework score: "))

    # Attendance with up to 3 attempts
    attendance_pct = -1
    for attempt in range(3):
        attendance_pct = float(input("Enter your attendance percentage score: "))
        if attendance_pct < 0:
            print("  Invalid attendance -- try again.")
        else:
            break
    else:
        print("Skipping student due to invalid attendance entries.")
        continue

    # Compute final score (same weighting as subtask 4)
    final_score = exam_score * 0.6 + cw_score * 0.4

    if not user_name:
        user_name = "Anonymous Student"

    # Grade classification
    if final_score >= 90:
        grade = "A"
    elif final_score >= 80:
        grade = "B"
    elif final_score >= 70:
        grade = "C"
    elif final_score >= 60:
        grade = "D"
    else:
        grade = "F"

    # PASS/FAIL
    if final_score >= 50 and attendance_pct >= 75:
        result = "PASS"
    else:
        result = "FAIL"

    # Update running totals
    student_counter += 1
    sum_final_score += final_score
    if result == "PASS":
        pass_counter += 1

    # Print student report line
    print(student_counter, ". ", user_name, " | ", "Final: ", round(final_score, 1), " | ", "Grade: ", grade, " | ", result, sep="")

# After loop ends, print class summary
print("="*40)
if student_counter > 0:
    class_average = sum_final_score / student_counter
    print("Class average  :", round(class_average, 1))
    print("Students passed:", pass_counter, "/", student_counter)
else:
    print("No students were entered.")
print("="*40)
````
