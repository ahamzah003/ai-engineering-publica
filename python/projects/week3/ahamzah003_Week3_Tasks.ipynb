## Task 1: Word Counter (Easy)

Ask for a sentence. Make it lowercase and split it into words. Count how many times each word appears using a **dictionary** (use `.get()`), and print each word with its count using `.items()`. Then use a **set** to print how many unique words there are.

**Expected Output** (example: sentence `"The cat and the dog and the bird"`):
```
the : 3
cat : 1
and : 2
dog : 1
bird : 1
Unique words: 5
```

````python
# Ask for a sentence
sentence = input("Enter a sentence: ")

# Convert to lowercase and split into words
words = sentence.lower().split()

# Dictionary to hold word counts
word_counts = {}

# Count each word using .get()
for word in words:
    word_counts[word] = word_counts.get(word, 0) + 1

# Print each word with its count using .items()
for word, count in word_counts.items():
    print(f"{word} : {count}")

# Use a set to find unique words
unique_words = set(words)
print("Unique words:", len(unique_words))

````

## Task 2: Three Little Functions (Easy)

Write three functions and call each one:
1. `area_of_rectangle(length, width)` returns the area
2. `is_even(number)` returns `True` or `False`
3. `greet(name, greeting="Hello")` returns a greeting like `"Hello, Ada!"`. The greeting has a **default value**

Use `return` (not `print`) inside the functions, and print the results when you call them.

**Expected Output:**
```
Area: 15
Is 7 even? False
Is 10 even? True
Hello, Ada!
Good morning, Bola!
```

````python
# Function to calculate the area of a rectangle
def area_of_rectangle(length, width):
    return length * width

# Function to check if a number is even
def is_even(number):
    return number % 2 == 0

# Function to greet someone, with a default greeting
def greet(name, greeting="Hello"):
    return f"{greeting}, {name}!"

# Call the functions and print results
print("Area:", area_of_rectangle(10, 3))
print("Is 100001 even?", is_even(100001))
print("Is 100020 even?", is_even(100020))
print(greet("Jack"))
print(greet("Bayo", "Good morning"))

````

## Task 3: Safe Division (Medium)

Ask for two numbers and divide the first by the second. Use `try` / `except` so the program never crashes: catch `ValueError` (not a number) and `ZeroDivisionError` with a friendly message each. Use `else` to print the result and `finally` to always print `Done.`. Test it with good numbers, zero, and text.

**Expected Output** (example: `8` and `0`):
```
You cannot divide by zero.
Done.
```

````python
try:
    # Ask for two numbers
    num1 = float(input("Enter first number: "))
    num2 = float(input("Enter second number: "))

    # Attempt division
    result = num1 / num2

except ValueError:
    # Handles non-numeric input
    print("That was not a valid number.")

except ZeroDivisionError:
    # Handles division by zero
    print("You cannot divide by zero.")

else:
    # Runs only if no error occurred
    print("Result:", result)

finally:
    # Always runs, no matter what
    print("Done.")

````

## Task 4: Circle & Calendar (Medium)

Use `import math` and `from datetime import date`. Ask for a radius and print the circle's area (`math.pi * radius ** 2`, rounded to 2 decimals), the area rounded up (`math.ceil`) and down (`math.floor`), and the square root of the area (`math.sqrt`). Then create `date(2026, 1, 1)` and `date(2026, 12, 25)`, and print the number of days between them (`.days`) and the weekday name of the second date (`.strftime("%A")`).

**Expected Output** (example: radius `5`):
```
Area         : 78.54
Rounded up   : 79
Rounded down : 78
Square root  : 8.86
Days between : 358
25 Dec 2026  : Friday
```

````python
import math
from datetime import date

# Ask for radius
radius = float(input("Enter radius: "))

# Circle area
area = math.pi * radius ** 2
print("Area         :", round(area, 2))

# Rounded up and down
print("Rounded up   :", math.ceil(area))
print("Rounded down :", math.floor(area))

# Square root of area
print("Square root  :", round(math.sqrt(area), 2))

# Dates
start_date = date(2026, 1, 1)
end_date = date(2026, 12, 25)

# Days between
days_between = (end_date - start_date).days
print("Days between :", days_between)

# Weekday name of second date
print("25 Dec 2026  :", end_date.strftime("%A"))

````

## Task 5: Gradebook (Challenge)

Build a small gradebook that keeps students and scores in a **dictionary**.
1. Write `check_score(score)` that raises a `ValueError` (with a message) if the score is below 0 or above 100, otherwise returns it
2. Write `get_grade(score)` that returns `"A"` (70+), `"B"` (60+), `"C"` (50+) or `"F"`
3. Loop: ask for a student name (stop on `done`) and a score. Wrap the score in `try` / `except ValueError` so a bad score prints `Invalid score:` and the error message, then `continue`
4. Print each student, score and grade using `.items()`, then the average (use `math.fsum`) and the highest score

**Expected Output** (example: `Ada` `80`; `Bola` `abc`; `Chidi` `105`; `Chidi` `67`; then `done`):
```
Invalid score: could not convert string to float: 'abc'
Invalid score: Score must be between 0 and 100.
Ada       80.0  A
Chidi     67.0  B
Average: 73.5
Highest: 80.0
```

````python
# for handling mathematical operations
import math

# Function to validate score
def check_score(score):
    if score < 0 or score > 100:
        raise ValueError("Score must be between 0 and 100.")
    return score

# Function to convert score to grade
def get_grade(score):
    if score >= 70:
        return "A"
    elif score >= 60:
        return "B"
    elif score >= 50:
        return "C"
    else:
        return "F"

# Dictionary to hold students and scores
gradebook = {}

# Input loop
while True:
    name = input("Enter student name (or 'done' to finish): ").strip().title()
    if name.lower() == "done":
        break

    try:
        score = float(input(f"Enter score for {name}: "))
        score = check_score(score)
    except ValueError as e:
        print("Invalid score:", e)
        continue
    else:
        gradebook[name] = score

# Print results
for student, score in gradebook.items():
    print(f"{student:<10} {score:.1f}  {get_grade(score)}")

# Average using math.fsum
# Avoid calculation if an empty dictionary
if gradebook:
    scores = list(gradebook.values())
    average = math.fsum(scores) / len(scores)
    highest = max(scores)
    print("Average:", round(average, 1))
    print("Highest:", round(highest, 1))
````
